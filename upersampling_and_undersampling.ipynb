```python
import numpy as np
import pandas as pd

np.random.seed(1) #for reproducibility

no_samples = 1000
class_0_ratio = 0.9
no_class_0 = int(no_samples * class_0_ratio)
no_class_1 = 100
```

```python
class_0 = {'feature1': np.random.normal(0, 1, no_class_0),
          'feature2': np.random.normal(0, 1, no_class_0),
          'target': [0]*no_class_0}
```

```python
print(class_0)
```

```python
class_1 = pd.DataFrame({'feature1': np.random.normal(3, 1, no_class_1),
          'feature2': np.random.normal(3, 1, no_class_1),
          'target': [1]*no_class_1})
```

```python
class_0=pd.DataFrame(class_0)
class_1=pd.DataFrame(class_1)
```

```python
df=pd.concat([class_0,class_1])
```

```python
df
```

```python
#upsampling
```

**upsampling**

```python
from sklearn.utils import resample
df_minority = df[df.target == 1]
df_majority = df[df.target == 0]
```

```python
df_minority_upsampled=resample(df_minority,n_samples=len(df_majority),replace=True,random_state=1)
```

```python
df_minority_upsampled
```

```python
df_upsampled=pd.concat([df_minority_upsampled,df_majority])
df_upsampled
```

```python
df_upsampled.target.value_counts()
```

#undersampling

```python
df_majority_downsampling=resample(df_majority,replace=False,n_samples=len(df_minority),random_state=1)
df_undersampling=pd.concat([df_majority_downsampling,df_minority])
df_undersampling
```

```python
df_undersampling.target.value_counts()
```
