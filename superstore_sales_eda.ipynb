# Outlier Detection & Handling (Boxplots & IQR Method)

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df.info()
```

```python
df.describe()
```

```python
#iqr=q3-q1
Q1=df["Sales"].quantile(0.25)
Q3=df["Sales"].quantile(0.75)
IQR=Q3-Q1
lower_bound=Q1-1.5*IQR
upper_bound=Q3+1.5*IQR
df_cleaned = df[(df['Sales'] >= lower_bound) & (df['Sales'] <= upper_bound)]
#df_cleaned=df[(df["Sales"]>=lower_bound) and  (df["Sales"]<=upper_bound)]  # give error
df_cleaned
```

# Univariate Analysis (Histograms & Bar Plots)
Why do we perform Univariate Analysis?

We want to answer questions like:

What is the average value?

What is the minimum and maximum?

Is the data normally distributed?

Are there any outliers?

How are the values spread?

Are there missing values?


```python
df_cleaned["Sales"].value_counts()
```

```python
plt.figure(figsize=(10,5))
sns.histplot(df_cleaned['Sales'], bins=30, kde=True, color='purple')
plt.title('Sales Distribution')
plt.show()

print(": Sales distribution is right-skewed. The vast majority of customer orders are small transactions (under $50), while high-value transactions above $200 are relatively rare.")
```

```python
#  Categorical Column Frequency (sns.countplot())
#Text columns (jaise Category ya Region) ki frequencies counting display karne ke liye Countplot sabse best hota hai.
plt.figure(figsize=(8, 4))
sns.countplot(data=df_cleaned, x='Category', hue='Category', palette='Set2', legend=False)
plt.title('Count of Orders by Category')
plt.xlabel('Category')
plt.ylabel('Total Orders')
plt.show()
print("Conclusion: Office Supplies is the most frequently ordered category by a huge margin (over 5,000 orders), whereas Furniture and Technology have significantly lower order volume (around 1,500 each).")
```

# Bivariate Analysis (Scatter Plots & Discount vs Profit)

```python
df_cleaned.info()
```

```python
df_cleaned.head()
```

```python
# 1. First make an explicit copy so Pandas stops giving warnings
df_clean_dates = df_cleaned.copy()

# 2. Re-parse dates
df_clean_dates['Order Date'] = pd.to_datetime(df_clean_dates['Order Date'], dayfirst=True, format='mixed')
df_clean_dates['Ship Date'] = pd.to_datetime(df_clean_dates['Ship Date'], dayfirst=True, format='mixed')

# 3. Calculate Shipping Days
df_clean_dates['Shipping Days'] = (df_clean_dates['Ship Date'] - df_clean_dates['Order Date']).dt.days

# 4. Filter only valid shipping days (Removing Negative values)
df_final = df_clean_dates[df_clean_dates['Shipping Days'] >= 0].copy()

# 5. Check Output
print(f"Total valid rows left: {len(df_final)}")
```

```python
df_final.head()
```

```python
# 1. Scatter Plot: Sales vs Shipping Days
plt.figure(figsize=(9, 5))
sns.scatterplot(data=df_final, x='Shipping Days', y='Sales', hue='Category', alpha=0.7)
plt.title('Sales vs Shipping Days by Category')
plt.xlabel('Shipping Days')
plt.ylabel('Sales ($)')
plt.show()
print("After date cleaning, all shipping durations are valid ($0+$ days) with most orders dispatched within 0–10 days. Shipping time is uniform across all product categories.")
```

```python
plt.figure(figsize=(9, 5))
sns.barplot(data=df_cleaned, x='Category', y='Sales', estimator=sum, palette='Set2', hue='Category', legend=False)
plt.title('Total Sales by Category')
plt.xlabel('Category')
plt.ylabel('Total Sales ($)')
plt.show()
print("Office Supplies dominates total sales revenue ($330K+) driven by high order volume, followed by Furniture ($240K) and Technology ($225K).")
```

# Monthly Sales Trend Calculate Karna (groupby())

```python
# Order Date ko sort karke Year-Month according total sales group karenge
df_monthly = df_final.groupby(['Order Year', 'Order Month'])['Sales'].sum().reset_index()

# Year aur Month ko combine karke ek readable Date string banayein (e.g., "2017-08")
df_monthly['YearMonth'] = df_monthly['Order Year'].astype(str) + '-' + df_monthly['Order Month'].astype(str).str.zfill(2)
```

```python
df_monthly=df_final.groupby(['order year','order month'])['Sales'].sum().reset_index()
```

```python
plt.figure(figsize=(14, 5))
sns.lineplot(data=df_monthly, x='YearMonth', y='Sales', marker='o', color='blue', linewidth=2)
plt.xticks(rotation=45)
plt.title('Monthly Sales Trend Over Time')
plt.xlabel('Year-Month')
plt.ylabel('Total Sales ($)')
plt.grid(True)
plt.show()

print("  1. Business Growth (Line Upar Ja Rahi Hai): Sales saal-dar-saal continuously badh rahi hai (2015 mein average sales $10,000 thi, jo 2018 tak badhkar $20,000+ ho gayi)",)

print("2. Festive / Year-End Season Peak: Har saal ke aakhri mahino mein (September se December) sales bohot high hoti hai, jabki January mein sabse low hoti hai.")

print("3. Record Sales: Pure dataset mein sabse highest sales 2018 ke February aur August mein touch hui ($25,000+).")
```
