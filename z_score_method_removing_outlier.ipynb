```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
salary = [11, 40, 45, 68, 65, 68, 78, 90, 57, 74, 91, 92, 88, 68, 57, 48, 99, 101, 68, 77, 110, 140]
df=pd.DataFrame(salary,columns=["salary"])
df
```

```python
sns.boxplot(df["salary"])
```

 **finding outlier using z_ score method**

---






```python
u=df["salary"].mean()
std=df["salary"].std()
print(f"mean of salry column is :{u} and std_deviation is : {std}")
```

```python
#calculate z_score
z_score=(df["salary"]-u)/std
print(z_score)
```

```python
df["z_score"]=z_score # add coulmn z_score in data frame so that we find  z > 3 and z <-3  row
```

```python
# Identify outliers where Z-score is greater than 3 or less than -3( but here we take 2.4 and 2.5 )
outliers = df[(df['z_score'] > 2.4) | (df['z_score'] < -2.5)]

# Display the outlier rows
print(outliers)
```

```python

 # remove outlier row from data frame

col=["salary","z_score"]
df_clean=df[~((df['z_score'] > 2.4) | (df['z_score'] < -2.5))]
df_clean
```

```python
# box plot  after  removing outlier
sns.boxplot(df_clean["salary"])
```
