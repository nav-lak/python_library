```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.DataFrame({"salary" : [18000, 20000, 22000, 21000, 23000, 25000, 30000, 100000]})
print("Before Capping:")
print(df)
```

```python
print("box plot before capping")
sns.boxplot(df["salary"])
```

```python
lower_cap = df["salary"].quantile(0.05)
upper_cap = df["salary"].quantile(0.95)

print("Lower cap (5%):", lower_cap)
print("Upper cap (95%):", upper_cap)
```

```python
# 3. Apply capping using np.where
# --------------------------------
df["salary"] = np.where(
    df["salary"] < lower_cap, lower_cap,
    np.where(df["salary"] > upper_cap, upper_cap, df["salary"])
)

print("After Capping:")
print(df)
```

```python
print("box plot after capping")
sns.boxplot(df["salary"])
```

```python
# apply again capping
lower_cap = df["salary"].quantile(0.05)
upper_cap = df["salary"].quantile(0.85)
print("Lower cap (5%):", lower_cap)
print("Upper cap (95%):", upper_cap)
# 3. Apply capping using np.where
# --------------------------------
df["salary"] = np.where(
    df["salary"] < lower_cap, lower_cap,
    np.where(df["salary"] > upper_cap, upper_cap, df["salary"])
)

print("After Capping:")
print(df)
```

```python
print("box plot after again  capping")
plt.xlabel("box plot after again  capping")
sns.boxplot(df["salary"])
```
